# Using Opik with OpenAI Agents

This notebook demonstrates how to capture and visualize traces from agent workflows using Opik.

```
import os
from opik import Opik

# Configuration
os.environ["OPIK_API_KEY"] = "your_api_key_here"
opik = Opik()
```

## Agent Workflow

Here is a simple example of an agent capturing its thought process.

```
# Example agent logic
def agent_workflow(user_input):
    with opik.trace(name="agent_workflow"):
        # Simulate agent thinking
        print(f"Processing: {user_input}")
        return "Success"

agent_workflow("How do I integrate Opik with my agent?")
```

## Visualizing Traces

Once the agent runs, you can see the full trace in the Opik dashboard.
